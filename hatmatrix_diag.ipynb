```python
import numpy as np
import sympy as sp
```

```python
def symbolic_limit_power(P):
    n = P.rows
    I = sp.eye(n)

    null_r = (P - I).nullspace()
    null_l = (P.T - I).nullspace()

    if len(null_r) != 1 or len(null_l) != 1:
        raise ValueError("Expected a 1-dimensional eigenspace for eigenvalue 1.")

    v = null_r[0]
    w = null_l[0]

    denom = sp.simplify((w.T * v)[0])
    if denom == 0:
        raise ValueError("w^T v = 0: eigenvalue 1 is defective, so P^n does not converge.")

    return sp.simplify((v * w.T) / denom)
```

```python
r = sp.symbols('r')
P_hat = sp.Matrix([[0,       2*r/(1 + 2*r), 0,       1/(1 + 2*r)],
                   [1/(3+r), 0,             r/(3+r), 0          ],
                   [0,       2/3,           0,       0          ],
                   [0,       0,             0,       1          ]])

P_hat_inf = symbolic_limit_power(P_hat)
```

```python
V, D = P_hat.diagonalize()
```

```python
D[2, 2] = 0
D[3, 3] = 0
D
```

```python
V @ D @ V.inv()  # This should give us P_hat_inf
```
